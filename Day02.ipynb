simple python program to find the total savings



```python
income=500000
food=25000
transport=12000
entertainment=10000
other_expenses=60000

total_expenses=food+transport+entertainment+other_expenses
savings=income-total_expenses

print("======================================")
print("Total Savings : ",savings,"LKR")
```

**Input The Values or User Inputs In Python**

```python
income = float(input("Enter The Income : "))
food = float(input("Enter The Food Expenses : "))
transport = float(input("Enter The Transport Expenses : "))
entertainment = float(input("Enter The Entertainment Expenses : "))
other_expenses = float(input("Enter The Other Expenses : "))

total_expenses = food + transport + entertainment + other_expenses
savings = income - total_expenses

print("======================================")
print("Total Savings : ", savings, "LKR")
```
